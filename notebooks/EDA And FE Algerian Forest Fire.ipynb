### Algerian Forest Fires Dataset

Data Set Information

The dataset includes 244 instances that regroup a data of two regions of Algeria, namely the Bejaia region located in the northeast of Algeria and the Sidi Bel-abbes region located in the northeast of Algeria.

122 instances for each region.

The period from June 2012 to September 2012. The dataset includes 11 attributes ans 1 output attributes (class) The 244 instances have been classified into fire(138 classes) and not fire (106 classes) classes.

Attributes Information:

1.Date : (DD/MM/YYYY)Day,Month(June to September), Year(2012) Weather data observations

2.Temp : Temperature noon(Temperature max) in Celsius degrees: 22 to 42

3.RH : Relative Humidity in % : 21 to 90

4.Ws : Wind speed in km/h : 6 to 29

5.Rain : Total day in mm: 0 to 16.8 FWI Components

6.Fine Fuel Moisture Code(FFMC) index from the FWI system: 28.6 to 92.5

7.Duff Moisture Code(DMC) index from the FWI system : 1.1 to 65.9

8.Drought Code(DC) index from the FWI system :7 to 220.4

9.Initial Spread Index(ISI) index from the FWI System : 0 to 18.5

10.Building Index(BUI) index from the FWI System : 1.1 to 68

11.Fire Weather Index(FWI) index : 0 to 31.1

12.Classes: Two Classes, namely Fire and Not Fire

### Data Cleaning

```python
## missing values
dataset[dataset.isnull().any(axis=1)]
```

The dataset is converted into two sets based on Region from 122th index, we can make a new column based on the Region

1."Bejaia Region Dataset"

2."Sidi-Bel Abbes Region Dataset"

Add new column with Region

```python
# Assign Region: 0 = Bejaia (rows 0-121), 1 = Sidi-Bel Abbes (rows 123-246)
# Row 122 is a separator/non-data row, will be dropped below
df = dataset.copy()
df.loc[:121, 'Region'] = 0   # Bejaia Region
df.loc[123:, 'Region'] = 1   # Sidi-Bel Abbes Region

# Drop the separator row (122) and reset the index
df = df.dropna(subset=['Region'])
df = df.reset_index(drop=True)
```

```python
df.info()
```

```python
df.head()
```

```python
# Drop rows with NaN in Region column first, then convert to int
df = df.dropna(subset=['Region'])
df['Region'] = df['Region'].astype(int)
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
## Removing the null values
df=df.dropna().reset_index(drop=True)
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
# Show the boundary rows between the two regions
# Index 121 = last Bejaia row (Region 0)
# Index 122 = first Sidi-Bel Abbes row (Region 1)
display(df.iloc[[121, 122]])
```

```python
## remove the 122nd row
df=df.drop(index=122).reset_index(drop=True)
```

```python
df.iloc[[122]]
```

```python
df.columns
```

```python
## Fix spaces in columns names
df.columns=df.columns.str.strip()
df.columns
```

```python
df.info()
```

#### Changes the required columns as integer datatype

```python
df[['month', 'day', 'year','Temperature','Ws',"RH"]]=df[['month', 'day', 'year','Temperature','Ws',"RH"]].astype(int)
```

```python
df.info()
```
